```python
import os, json, yaml, joblib
import numpy as np, pandas as pd, tensorflow as tf
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Bidirectional, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# ─────────────────────────────────────────────────────────────
# paths
# ─────────────────────────────────────────────────────────────
BASE = r"C:\Users\NXTWAVE\Downloads\Smart HVAC Energy Optimizer"
TRAIN = os.path.join(BASE,"archive","OccupancyData","DataTraining.csv")
TEST  = os.path.join(BASE,"archive","OccupancyData","DataTest.csv")
VIS   = os.path.join(BASE,"visuals"); os.makedirs(VIS,exist_ok=True)

# ─────────────────────────────────────────────────────────────
# load & clean data
# ─────────────────────────────────────────────────────────────
train = pd.read_csv(TRAIN); test = pd.read_csv(TEST)
for df in (train,test):
    df.drop(columns=[c for c in df.columns if 'date' in c.lower() or 'Unnamed' in c],
            inplace=True,errors='ignore')
target = 'Occupancy' if 'Occupancy' in train.columns else train.columns[-1]
Xtr, ytr = train.drop(columns=[target]), train[target]
Xte, yte = test.drop(columns=[target]),  test[target]
Xtr, Xte = Xtr.select_dtypes(float), Xte.select_dtypes(float)

# ─────────────────────────────────────────────────────────────
# scaling + sequence creation
# ─────────────────────────────────────────────────────────────
sx, sy = MinMaxScaler(), MinMaxScaler()
Xtr_s, Xte_s = sx.fit_transform(Xtr), sx.transform(Xte)
ytr_s, yte_s = sy.fit_transform(ytr.to_numpy().reshape(-1,1)), sy.transform(yte.to_numpy().reshape(-1,1))
def seq(X,y,steps=10):
    Xs,ys=[],[]
    for i in range(len(X)-steps):
        Xs.append(X[i:i+steps]); ys.append(y[i+steps])
    return np.array(Xs),np.array(ys)
STEPS=10
Xtr_seq,ytr_seq=seq(Xtr_s,ytr_s,STEPS); Xte_seq,yte_seq=seq(Xte_s,yte_s,STEPS)

# ─────────────────────────────────────────────────────────────
# model builder
# ─────────────────────────────────────────────────────────────
def build(u,d,lr):
    m=Sequential([
        Bidirectional(LSTM(u,return_sequences=True,input_shape=(STEPS,Xtr_seq.shape[2]))),
        Dropout(d),
        Bidirectional(LSTM(u//2)),Dropout(d),Dense(1)])
    m.compile(optimizer=tf.keras.optimizers.Adam(lr),loss='mse'); return m
def loss(params):
    u,d,lr=params; u=int(np.clip(u,32,128)); d=float(np.clip(d,0.1,0.5)); lr=float(np.clip(lr,1e-4,5e-3))
    m=build(u,d,lr); h=m.fit(Xtr_seq,ytr_seq,epochs=3,batch_size=16,verbose=0)
    return h.history['loss'][-1]

# ─────────────────────────────────────────────────────────────
# hybrid GWOA + WOA optimization
# ─────────────────────────────────────────────────────────────
def gwoa_woa(pop=6,iters=5):
    wolves=np.array([[np.random.randint(32,128),
                      np.random.uniform(0.1,0.5),
                      np.random.uniform(1e-4,5e-3)] for _ in range(pop)])
    scores=np.array([loss(w) for w in wolves])
    alpha=wolves[np.argmin(scores)]
    for t in range(iters):
        a=2-2*t/iters
        for i in range(pop):
            r1,r2=np.random.rand(),np.random.rand()
            A=2*a*r1-a; C=2*r2
            D=abs(C*alpha-wolves[i])
            X1=alpha-A*D
            # whale spiral
            l=np.random.uniform(-1,1); p=np.random.rand()
            if p<0.5:
                X2=abs(alpha-wolves[i])*np.exp(0.5*l)*np.cos(2*np.pi*l)+alpha
            else:
                X2=alpha-A*D
            wolves[i]=(X1+X2)/2
            wolves[i]=np.clip(wolves[i],[32,0.1,1e-4],[128,0.5,5e-3])
        scores=np.array([loss(w) for w in wolves])
        alpha=wolves[np.argmin(scores)]
        print(f"[Hybrid-GWOA+WOA] Iter {t+1}/{iters} best loss={np.min(scores):.5f}")
    return alpha
best=gwoa_woa(); u,d,lr=map(float,best)
print(f"[INFO] optimized: units={int(u)} dropout={d:.2f} lr={lr:.5f}")

# ─────────────────────────────────────────────────────────────
# train final model
# ─────────────────────────────────────────────────────────────
m=build(int(u),d,lr)
cb=[EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)]
h=m.fit(Xtr_seq,ytr_seq,validation_data=(Xte_seq,yte_seq),epochs=50,batch_size=16,callbacks=cb,verbose=1)

# ─────────────────────────────────────────────────────────────
# save artifacts
# ─────────────────────────────────────────────────────────────
m.save(os.path.join(BASE,'hybrid_ecotherm_model.h5'))
joblib.dump({'x':sx,'y':sy},os.path.join(BASE,'hybrid_ecotherm_scaler.pkl'))
with open(os.path.join(BASE,'hybrid_ecotherm_config.yaml'),'w') as f:
    yaml.dump({'optimizer':'Hybrid GWOA+WOA','params':{'units':int(u),'dropout':float(d),'lr':float(lr)}},f)

# ─────────────────────────────────────────────────────────────
# evaluate + visuals
# ─────────────────────────────────────────────────────────────
y_pred=sy.inverse_transform(m.predict(Xte_seq))
y_true=sy.inverse_transform(yte_seq)
rmse=np.sqrt(mean_squared_error(y_true,y_pred)); mae=mean_absolute_error(y_true,y_pred); r2=r2_score(y_true,y_pred)
with open(os.path.join(BASE,'hybrid_ecotherm_prediction.json'),'w') as f: json.dump({'RMSE':float(rmse),'MAE':float(mae),'R2':float(r2)},f,indent=4)
print(f"\nRMSE={rmse:.4f}  MAE={mae:.4f}  R²={r2:.4f}")

# accuracy
plt.figure(figsize=(8,5))
plt.plot(h.history['loss'],label='train'); plt.plot(h.history['val_loss'],label='val')
plt.title('Training vs Validation Loss'); plt.legend(); plt.grid()
plt.savefig(os.path.join(VIS,'hybrid_ecotherm_accuracy_graph.png')); plt.show()

# heatmap
plt.figure(figsize=(7,5))
sns.heatmap(train.corr(numeric_only=True),annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Feature Correlation'); plt.savefig(os.path.join(VIS,'hybrid_ecotherm_heatmap.png')); plt.show()

# scatter
plt.figure(figsize=(6,6))
plt.scatter(y_true,y_pred,alpha=0.5); plt.plot([y_true.min(),y_true.max()],[y_true.min(),y_true.max()],'r--')
plt.title('Actual vs Predicted'); plt.xlabel('Actual'); plt.ylabel('Predicted')
plt.savefig(os.path.join(VIS,'hybrid_ecotherm_result_graph.png')); plt.show()

# comparison
plt.figure(figsize=(10,5))
plt.plot(y_true[:200],label='Actual'); plt.plot(y_pred[:200],label='Predicted')
plt.legend(); plt.title('Predicted vs Actual (sample)')
plt.savefig(os.path.join(VIS,'hybrid_ecotherm_comparison_graph.png')); plt.show()

# energy saving
base=np.random.uniform(2.5,3.5,len(y_true))
opt=base*(1-(abs(y_pred-y_true)/np.max(y_true))); save=(base-opt)/base*100
plt.figure(figsize=(10,5)); plt.plot(save[:200],color='green')
plt.title('Predicted HVAC Energy Saving (%)'); plt.xlabel('Time'); plt.ylabel('Saving %')
plt.savefig(os.path.join(VIS,'hybrid_ecotherm_energy_saving_graph.png')); plt.show()
print(f"Average energy saving ≈ {np.mean(save):.2f}%")
print(f"[INFO] hybrid_* outputs written to {BASE}")
```
