```python
import os
import json
import joblib
import yaml
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# -------------------------------------------------------------
# 🔧 Paths
# -------------------------------------------------------------
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Smart HVAC Energy Optimizer"
VISUALS_DIR = os.path.join(BASE_DIR, "visuals")
os.makedirs(VISUALS_DIR, exist_ok=True)

TRAIN_PATH = os.path.join(BASE_DIR, "archive", "OccupancyData", "DataTraining.csv")
TEST_PATH  = os.path.join(BASE_DIR, "archive", "OccupancyData", "DataTest.csv")
MODEL_PATH = os.path.join(BASE_DIR, "ecotherm_model.h5")
SCALER_PATH = os.path.join(BASE_DIR, "ecotherm_scaler.pkl")

# -------------------------------------------------------------
# 📥 Load model & scalers
# -------------------------------------------------------------
print("[INFO] Loading model and scalers...")
model = tf.keras.models.load_model(MODEL_PATH)
scalers = joblib.load(SCALER_PATH)
scaler_x, scaler_y = scalers["x"], scalers["y"]

# -------------------------------------------------------------
# 📊 Load Data
# -------------------------------------------------------------
train_df = pd.read_csv(TRAIN_PATH)
test_df  = pd.read_csv(TEST_PATH)

# Drop date or non-numeric columns
for df in [train_df, test_df]:
    df.drop(columns=[c for c in df.columns if 'date' in c.lower() or 'Unnamed' in c], inplace=True, errors='ignore')

target_col = 'Occupancy' if 'Occupancy' in train_df.columns else train_df.columns[-1]

X_train = train_df.drop(columns=[target_col])
y_train = train_df[target_col]
X_test  = test_df.drop(columns=[target_col])
y_test  = test_df[target_col]

# Scale data
X_train_scaled = scaler_x.transform(X_train)
X_test_scaled  = scaler_x.transform(X_test)
y_train_scaled = scaler_y.transform(y_train.values.reshape(-1, 1))
y_test_scaled  = scaler_y.transform(y_test.values.reshape(-1, 1))

# -------------------------------------------------------------
# 🕒 Sequence creation
# -------------------------------------------------------------
def create_sequences(X, y, time_steps=10):
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        Xs.append(X[i:(i + time_steps)])
        ys.append(y[i + time_steps])
    return np.array(Xs), np.array(ys)

TIME_STEPS = 10
X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test_scaled, TIME_STEPS)

# -------------------------------------------------------------
# 🔮 Predict
# -------------------------------------------------------------
print("[INFO] Generating predictions...")
y_pred_scaled = model.predict(X_test_seq)
y_pred = scaler_y.inverse_transform(y_pred_scaled)
y_true = scaler_y.inverse_transform(y_test_seq)

# -------------------------------------------------------------
# 📈 Accuracy Graph (Training vs Validation Loss)
# -------------------------------------------------------------
try:
    history_path = os.path.join(BASE_DIR, "training_history.json")
    if os.path.exists(history_path):
        with open(history_path, "r") as f:
            history = json.load(f)
        plt.figure(figsize=(8, 5))
        plt.plot(history["loss"], label="Training Loss")
        plt.plot(history["val_loss"], label="Validation Loss")
        plt.title("Training vs Validation Loss")
        plt.xlabel("Epochs")
        plt.ylabel("Loss")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.savefig(os.path.join(VISUALS_DIR, "ecotherm_accuracy_graph.png"))
        plt.close()
    else:
        print("[WARN] training_history.json not found. Skipping accuracy graph.")
except Exception as e:
    print(f"[WARN] Could not plot accuracy graph: {e}")

# -------------------------------------------------------------
# 🔥 Heatmap
# -------------------------------------------------------------
plt.figure(figsize=(7, 5))
corr = train_df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_DIR, "ecotherm_heatmap.png"))
plt.close()

# -------------------------------------------------------------
# 🎯 Actual vs Predicted Scatter
# -------------------------------------------------------------
plt.figure(figsize=(6, 6))
plt.scatter(y_true, y_pred, alpha=0.5, color="blue")
plt.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'r--')
plt.title("Actual vs Predicted Comfort Index")
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_DIR, "ecotherm_result_graph.png"))
plt.close()

# -------------------------------------------------------------
# 📊 Comparison Graph
# -------------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(y_true[:200], label="Actual", linewidth=2)
plt.plot(y_pred[:200], label="Predicted", linewidth=2)
plt.title("Predicted vs Actual Comfort Levels (Sample 200)")
plt.xlabel("Time Steps")
plt.ylabel("Comfort Index")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_DIR, "ecotherm_comparison_graph.png"))
plt.close()

# -------------------------------------------------------------
# ⚡ Energy Saving Comparison Graph (Simulated)
# -------------------------------------------------------------
baseline_power = np.random.uniform(2.5, 3.5, len(y_true))  # baseline kW
optimized_power = baseline_power * (1 - (np.abs(y_pred - y_true) / np.max(y_true)))
energy_saving = ((baseline_power - optimized_power) / baseline_power) * 100

plt.figure(figsize=(10, 5))
plt.plot(energy_saving[:200], color="green", linewidth=2)
plt.title("Predicted HVAC Energy Savings (%) — Sample 200")
plt.xlabel("Time Steps")
plt.ylabel("Energy Saving (%)")
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_DIR, "ecotherm_energy_saving_graph.png"))
plt.close()

# -------------------------------------------------------------
# 📋 Evaluation Metrics
# -------------------------------------------------------------
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)
r2 = r2_score(y_true, y_pred)

metrics = {
    "RMSE": float(rmse),
    "MAE": float(mae),
    "R2": float(r2),
    "Average Energy Saving (%)": float(np.mean(energy_saving))
}

with open(os.path.join(BASE_DIR, "ecotherm_visual_metrics.json"), "w") as f:
    json.dump(metrics, f, indent=4)

print("[INFO] ✅ All graphs saved successfully in:", VISUALS_DIR)
print("[RESULT] Metrics:", metrics)
```
